## 2. The question: decision, action, cost of a wrong call

* **The Search Question:** Which organic search pages are experiencing true structural performance decay and require an immediate content refresh?
* **Unit of Analysis:** A single page / URL observed over a 30-day window.
* **Model Output:** A probabilistic decay score (0.0 to 1.0) indicating the likelihood that a page's performance decline is structural decay rather than seasonal noise.
* **The Action Taken:** The content team queues high-priority decaying pages into an automated revision workflow (updating metadata, revising outdated body copy, and adding internal links).
* **Cost of a Wrong Call:**
  * **False Positive (Flagging a healthy page):** Wastes copywriter time and budget editing content that didn't need fixing, risking ranking instability on currently performing pages.
  * **False Negative (Missing a dying page):** Allows key organic landing pages to bleed traffic and revenue silently to competitors for months.
* **Why ML Helps:** Hardcoded rules (e.g., `clicks_drop > 20%`) fail because they treat seasonal noise and algorithm updates identically. ML models evaluate multiple non-linear signals simultaneously (impressions, CTR, positions, age, and historic traffic) to separate true decay from temporary blips.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
import numpy as np

# Load data directly from raw GitHub URL to avoid local path errors
url = "https://raw.githubusercontent.com/Tokollo-server/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"

try:
    df = pd.read_csv(url)
except Exception:
    # Fallback path if working locally outside of Colab
    df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Create starter target label
df["is_declining_label"] = df["trend_direction"].str.lower().eq("down").astype(int)

# Calculate key evidence metrics
total_pages = len(df)
declining_pages = df["is_declining_label"].sum()
pct_declining = (declining_pages / total_pages) * 100

avg_imp_declining = df[df["is_declining_label"] == 1]["impressions_90d"].mean()
avg_imp_healthy = df[df["is_declining_label"] == 0]["impressions_90d"].mean()
avg_days_stale = df[df["is_declining_label"] == 1]["days_since_last_update"].mean()

print("--- STARTER DATASET EVIDENCE ---")
print(f"1. Total Dataset Size: {total_pages:,} pages")
print(f"2. Decaying Pages Rate: {declining_pages:,} pages ({pct_declining:.1f}% of dataset)")
print(f"3. Exposure at Risk: Decaying pages average {avg_imp_declining:,.0f} 90-day impressions (vs {avg_imp_healthy:,.0f} for healthy pages)")
print(f"4. Staleness Factor: Decaying pages have an average of {avg_days_stale:.1f} days since their last update")
```

## 4. Careful words: what I can and can't claim

* **What I CAN claim:** The model identifies pages demonstrating statistical indicators of structural decay based on historical search performance features (impressions, position drift, and click volume shifts).
* **What I CANNOT claim:** The model cannot prove *why* a page declined (e.g., competitor actions, Google core updates, or technical site issues) nor guarantee that updating a page will automatically restore lost search traffic.
